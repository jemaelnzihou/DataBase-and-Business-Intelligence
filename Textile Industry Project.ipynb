## 🔍 **1. What This Model Shows (Technical Interpretation)**

### 🎯 Star Schema Design with Fact-Dimension Logic:

* **Fact Tables (metrics & transactions):**

  * `kpi_tracking`: tracks baseline vs improved KPIs (Cycle Time, Defect Rate)
  * `cost_savings`: financial impact (estimated vs actual savings, ROI)
  * `employees`: operational and compensation data
  * `gdp_data`, `interest_rates`, `hospitalization_rates`: macroeconomic time series

* **Dimension Tables (context):**

  * `projects`: links all KPIs and cost records
  * `departments`: describes employee units

✅ Power BI relationships:

* **1-to-many joins** are visible via `project_id`, `department_id`, `year`, and `country`—ensuring normalized, query-efficient design.


## 🧠 **2. What Insights Can Be Derived?**

### 📈 Operational & Financial KPIs

* Evaluate how project initiatives (linked by `project_id`) improved:

  * **Cycle Time**
  * **Production Rate**
  * **Defect Rate**
* Calculate and visualize **ROI trends** and **actual vs estimated savings**

### 🌎 Economic Contextualization

* Compare internal performance with:

  * **GDP trends** (was growth aligned with increased savings?)
